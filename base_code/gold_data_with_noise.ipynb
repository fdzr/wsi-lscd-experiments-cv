```python
from pathlib import Path

import pandas as pd
from scipy.stats import spearmanr
```

```python
path_to_data = "../input/gold_data_point_{factor}/{dataset}"
GOLD_DATA = {
    "dwug_es": pd.read_csv("../test_data_es.csv", sep="\t"),
    "dwug_en": pd.read_csv("../test_data_en.csv", sep="\t"),
}
TARGET_WORDS = {
    "dwug_es": GOLD_DATA["dwug_es"]["word"].tolist(),
    "dwug_en": open("../dwug_en/target_words.txt").read().split("\n"),
}
GOLD_DATA_GRADED_CHANGE = {
    "dwug_es": GOLD_DATA["dwug_es"].set_index("word")["change_graded"].to_dict(),
    "dwug_en": GOLD_DATA["dwug_en"].set_index("lemma")["change_graded"].to_dict(),
}
# print(TARGET_WORDS["dwug_es"])
print(GOLD_DATA_GRADED_CHANGE["dwug_en"])
```

```python
for d in ["dwug_es", "dwug_en"]:
    stat = {}
    print(d)
    
    for factor in ["one", "four", "seven"]:
        print(f"  {factor}", end=" ")
        p = Path(path_to_data.format(factor=factor, dataset=d))
        
        assert p.exists() is True
        
        data = pd.read_csv(next(p.glob("*scores")))
        
        v1, v2 = [], []
        for tw in TARGET_WORDS[d]:
            data_filtered = data[data["word"] == tw]
            avg = data_filtered["score"].mean(axis=0)
            v1.append(GOLD_DATA_GRADED_CHANGE[d][tw])
            v2.append(-1 * avg)
            
        spr, _ =spearmanr(v1, v2)
        print(f"{spr}")
            
        
        
```
